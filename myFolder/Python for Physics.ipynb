```python
import matplotlib.pyplot as plt
import numpy as np
import math
def correctfunction(x):
    return 15 + 2.4*x -0.5*x**2 - 0.35*x**3
x_list=np.linspace(-20,20,21)
data0=np.array([x_list,correctfunction(x_list)]) # unnoisy data
data=np.array([data0[0]+0.1*(2*np.random.rand(21)-1),data0[1]+5*(2*np.random.rand(21)-1)]) #noisy data
datapoly=data[:,7:14] # Small part of data
print(data)
print(datapoly)
#y=a0+a1*x**1+a2*x*2+a3*x**3+a4*x*4+a5*x**5+a6*x**6
#This same for the 7 data points
#In matrix form y=xa
x=np.transpose(np.array([datapoly[0]**0,datapoly[0]**1,datapoly[0]**2,datapoly[0]**3,datapoly[0]**4,datapoly[0]**5,datapoly[0]**6]))
print(x)
y=np.array(datapoly[1,:])
print(y)
a=np.linalg.solve(x,y)
x_values=np.linspace(-20,20,901)
y_values=np.array(a[0]*x_values**0+a[1]*x_values**1+a[2]*x_values**2+a[3]*x_values**3+a[4]*x_values**4+a[5]*x_values*5+a[6]*x_values**6)
#plt.xlim(-1,1)
plt.ylim(-75,75)
plt.scatter(datapoly[0],datapoly[1])
plt.plot(x_values,y_values,"blue")
```
